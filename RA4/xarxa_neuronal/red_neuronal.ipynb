# Clasificación con Red Neuronal: Bajas de Personal en la Guerra Rusia-Ucrania


## Objetivo
Vamos a predecir si las bajas diarias de personal superan la media usando una red neuronal sencilla. Trabajaremos con datos reales del conflicto Rusia-Ucrania.


## 1. Cargar y preparar los datos
Primero importamos las librerías y preparamos los datos para el modelo. El objetivo es crear una variable que indique si las bajas diarias superan la media.


```python
# --- 1. Cargar y preparar los datos ---
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('./russian_ukraine_war/russia_losses_personnel.csv')
media_bajas = df['personnel'].diff().mean() 
df['bajas_diarias'] = df['personnel'].diff().fillna(0)
df['objetivo'] = (df['bajas_diarias'] > media_bajas).astype(int)

X = df[['day']].values
y = df['objetivo'].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.33, random_state=17)
```

## 2. Construir y entrenar la red neuronal
Ahora creamos una red neuronal sencilla con Keras. Usamos solo una característica de entrada para simplificar el ejemplo.


```python
# --- 2. Construir y entrenar la red neuronal ---
from keras import models, layers

INPUT_DIM = X_train.shape[1]
model = models.Sequential()
model.add(layers.Dense(4, activation='relu', input_dim=INPUT_DIM))
model.add(layers.Dense(2, activation='relu'))
model.add(layers.Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
print("Entrenando con tus datos de personal...")
model.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)
```

## 3. Evaluar el modelo
Finalmente, evaluamos la precisión del modelo con los datos de prueba. Así sabremos qué tan bien predice si las bajas diarias superan la media.


```python
# --- 3. Evaluar el modelo ---
loss, accuracy = model.evaluate(X_test, y_test)
print(f"\nPrecisión del modelo con tu dataset: {accuracy:.4f}")
```
